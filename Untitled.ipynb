4. Фильтрация

Оставляем только треки, выпущенные после 2024 года.

```python
tracks_after_2024 = df[df['release_year'] > 2024]

display(tracks_after_2024)
print("Количество треков после 2024 года:", tracks_after_2024.shape[0])
```

Основные статистические показатели

```python
pop_stats = df['popularity'].agg(['mean', 'median', 'min', 'max', 'std'])

display(pop_stats)

print(f"Среднее значение popularity: {pop_stats['mean']:.2f}")
print(f"Медиана popularity: {pop_stats['median']:.2f}")
print(f"Минимум popularity: {pop_stats['min']}")
print(f"Максимум popularity: {pop_stats['max']}")
print(f"Стандартное отклонение popularity: {pop_stats['std']:.2f}")
```

Мода

```python
genre_mode = df['genre'].mode()
year_mode = df['release_year'].mode()

print("Мода для genre:")
print(genre_mode)

print("\nМода для release_year:")
print(year_mode)
```

Гистограмма

```python
plt.figure(figsize=(10, 6))
plt.hist(df['popularity'], bins=20, color='skyblue', edgecolor='black')
plt.title('Распределение популярности треков')
plt.xlabel('Популярность')
plt.ylabel('Количество треков')
plt.grid(axis='y', alpha=0.3)
plt.show()
```

Дополнительно можно посчитать диапазон основной части значений:

```python
q1 = df['popularity'].quantile(0.25)
q3 = df['popularity'].quantile(0.75)

print(f"25-й процентиль: {q1}")
print(f"75-й процентиль: {q3}")
print(f"Основная часть значений примерно в диапазоне: {q1}–{q3}")
```

8. Ответы на контрольные вопросы

1. Что такое DataFrame?
DataFrame — это двумерная таблица данных в библиотеке Pandas. Она состоит из строк и столбцов, похожа на таблицу Excel или SQL-таблицу. В DataFrame могут храниться разные типы данных: числа, строки, даты, логические значения. У строк и столбцов есть индексы и названия. В данной работе DataFrame df содержит информацию о треках Spotify.
2. Что такое признак (Feature)?
Признак — это отдельная характеристика объекта, то есть столбец в таблице данных. Например, в нашем датасете признаками являются artist_name, genre, popularity, stream_count, release_year, tempo, duration_ms и другие. По признакам мы анализируем объекты: сравниваем, фильтруем, строим графики и ищем закономерности.
3. Чем среднее арифметическое отличается от медианы?
Среднее арифметическое — это сумма всех значений, делённая на их количество. Оно показывает «среднее» значение, но сильно зависит от выбросов.
Медиана — это значение, которое находится в середине упорядоченного ряда. Она делит данные так, что 50% значений меньше медианы и 50% больше. Медиана устойчива к выбросам.
Если данные распределены симметрично, среднее и медиана близки. Если есть сильно выделяющиеся значения, медиана лучше показывает типичное значение.
4. Что показывает гистограмма?
Гистограмма показывает распределение числового признака. По оси X откладываются интервалы значений, а по оси Y — частота или количество значений, попавших в каждый интервал. С помощью гистограммы можно увидеть:
    в каком диапазоне сосредоточена основная часть значений;
    есть ли пики и провалы;
    симметрично ли распределение;
    есть ли редкие значения и выбросы.
6. Для чего нужна фильтрация данных?
Фильтрация нужна для отбора нужных строк по определённому условию. Например, можно оставить только треки после 2024 года, только жанр Pop или только треки с популярностью выше 50. Фильтрация помогает анализировать не весь датасет сразу, а конкретную группу объектов, убирать лишние данные и находить закономерности внутри нужной части таблицы.
